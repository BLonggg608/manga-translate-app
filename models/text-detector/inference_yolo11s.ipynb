```python
from ultralytics import YOLO
from PIL import Image
import os
import requests

HOME = os.getcwd()

model = YOLO(f'{HOME}/training/runs/detect/train2/weights/best.pt')
image = Image.open(requests.get('https://static.wikia.nocookie.net/bluelock/images/2/21/Chapter_157.png/revision/latest/thumbnail/width/360/height/360?cb=20220327130008', stream=True).raw)
results = model.predict(image, conf=0.5, device='cuda:1', imgsz=1024)[0]
```
